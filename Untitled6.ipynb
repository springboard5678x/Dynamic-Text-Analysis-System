```python
#  imports + downloads
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pprint import pprint
from wordcloud import WordCloud
from tqdm.auto import tqdm

# nltk for stopwords
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords

# scikit-learn dataset and NMF
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer
from sklearn.decomposition import NMF

# spaCy for lemmatization
import spacy
nlp = spacy.load("en_core_web_sm", disable=["parser","ner"])

# gensim for LDA + coherence
import gensim
from gensim.utils import simple_preprocess
from gensim.corpora import Dictionary
from gensim.models.phrases import Phrases, Phraser
from gensim.models import CoherenceModel, LdaModel

# pyLDAvis (import later, only if needed to show interactive viz)
import warnings
warnings.filterwarnings("ignore", category=DeprecationWarning)
```

```python
#  fetch the 20 newsgroups dataset
data = fetch_20newsgroups(subset='all', remove=('headers','footers','quotes'), download_if_missing=True)
texts = data.data
targets = data.target
target_names = data.target_names

df = pd.DataFrame({"text": texts, "target": targets})
df['target_name'] = df['target'].apply(lambda x: target_names[x])

print("Total documents:", len(df))
print("Sample categories and counts:\n", df['target_name'].value_counts().head())
df.head(2)
```

```python
#  cleaning functions
STOPWORDS = set(stopwords.words('english'))

def clean_text(text):
    """Basic cleaning: remove emails, URLs, non-alpha, lowercasing, strip spaces."""
    text = str(text)
    text = re.sub(r'\S*@\S*\s?', '', text)         # remove emails
    text = re.sub(r'http\S+|www.\S+', '', text)    # remove URLs
    text = re.sub(r'[^a-zA-Z]', ' ', text)         # keep letters
    text = text.lower()
    text = re.sub(r'\s+', ' ', text).strip()
    return text

def lemmatize_texts(texts, nlp, allowed_postags=['NOUN','ADJ','VERB','ADV']):
    """
    Lemmatize list of texts using spaCy.
    Returns list-of-list tokens.
    """
    docs_lemmas = []
    for doc in nlp.pipe(texts, batch_size=200, disable=['parser','ner']):
        tokens = [token.lemma_ for token in doc
                  if token.pos_ in allowed_postags
                  and not token.is_stop
                  and token.lemma_ not in STOPWORDS
                  and len(token.lemma_) > 2]
        docs_lemmas.append(tokens)
    return docs_lemmas

# Apply cleaning + lemmatization
cleaned = [clean_text(t) for t in tqdm(df['text'].astype(str), desc="Cleaning")]
lemmatized_texts = lemmatize_texts(cleaned, nlp)

print("Example tokens (doc 0):", lemmatized_texts[0][:30])
```

```python
#  bigrams, dictionary and corpus
bigram = Phrases(lemmatized_texts, min_count=20, threshold=100)  # frequent phrase detector
bigram_mod = Phraser(bigram)

# Apply bigram model
texts_bigrams = [bigram_mod[doc] for doc in lemmatized_texts]

# Build gensim dictionary and corpus (BOW)
id2word = Dictionary(texts_bigrams)
# filter extremes to remove rare and overly frequent tokens
id2word.filter_extremes(no_below=15, no_above=0.5, keep_n=100000)
corpus = [id2word.doc2bow(text) for text in texts_bigrams]

print("Dictionary size:", len(id2word))
print("Example doc bow (first 20 tokens):", corpus[0][:20])
```

```python
#  model training helper function for LDA
from gensim.models import LdaModel
import math

def compute_lda_models(id2word, corpus, texts, start=5, limit=15, step=1,
                       passes=10, iterations=200, random_state=100):
    topics_range = list(range(start, limit+1, step))
    model_list = []
    coherence_values = []
    perplexity_values = []
    for num_topics in tqdm(topics_range, desc="Training LDA models"):
        model = LdaModel(corpus=corpus,
                         id2word=id2word,
                         num_topics=num_topics,
                         random_state=random_state,
                         update_every=1,
                         chunksize=2000,
                         passes=passes,
                         iterations=iterations,
                         alpha='auto',
                         per_word_topics=True)
        model_list.append(model)
        # coherence
        coherencemodel = CoherenceModel(model=model, texts=texts, dictionary=id2word, coherence='c_v')
        c_v = coherencemodel.get_coherence()
        coherence_values.append(c_v)
        # perplexity: gensim returns log bound; convert to perplexity
        bound = model.log_perplexity(corpus)
        # convert to approx perplexity; note: this is an approximate transform used for comparison
        perp = np.exp2(-bound)
        perplexity_values.append(perp)
    return topics_range, model_list, coherence_values, perplexity_values

# run for topics 5..15 (you can extend)
topics_range, model_list, coherence_values, perplexity_values = compute_lda_models(
    id2word=id2word, corpus=corpus, texts=texts_bigrams, start=5, limit=15, step=1,
    passes=10, iterations=200)
```

```python
#  plot curves
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.plot(topics_range, coherence_values, marker='o')
plt.xlabel("Number of topics")
plt.ylabel("Coherence (c_v)")
plt.title("Coherence vs #topics")
plt.axhline(0.6, color='red', linestyle='--', label='0.6 target')
plt.legend()

plt.subplot(1,2,2)
plt.plot(topics_range, perplexity_values, marker='o')
plt.xlabel("Number of topics")
plt.ylabel("Perplexity")
plt.title("Perplexity vs #topics (lower better)")

plt.tight_layout()
plt.show()

# Print best coherence
best_idx = int(np.argmax(coherence_values))
print(f"Best coherence at k={topics_range[best_idx]}: c_v={coherence_values[best_idx]:.4f}")
```

```python
#  show top topics for best LDA model
best_idx = int(np.argmax(coherence_values))
best_k = topics_range[best_idx]
best_lda = model_list[best_idx]
print(f"Selected LDA model: {best_k} topics, coherence={coherence_values[best_idx]:.4f}\n")
pprint(best_lda.print_topics(num_words=12))
```

```python
#  create wordclouds for each topic
def topic_wordcloud(model, id2word, topicnum, topn=50):
    """Return word frequency dict for a topic to feed WordCloud"""
    terms = model.get_topic_terms(topicid=topicnum, topn=topn)
    freqs = {id2word[id]: weight for id, weight in terms}
    return freqs

# Plot wordclouds in grid
cols = 3
rows = int(np.ceil(best_k / cols))
plt.figure(figsize=(15, 5 * rows))
for t in range(best_k):
    plt.subplot(rows, cols, t+1)
    freqs = topic_wordcloud(best_lda, id2word, t, topn=50)
    wc = WordCloud(width=600, height=400).generate_from_frequencies(freqs)
    plt.imshow(wc, interpolation='bilinear')
    plt.axis('off')
    plt.title(f"Topic {t}")
plt.tight_layout()
plt.show()
```

```python
#  NMF modeling using TF-IDF (works on vector space)
# We compute NMF for the same range of topic counts and compute coherence using gensim (from top words)
tfidf_vectorizer = TfidfVectorizer(max_df=0.5, min_df=15, stop_words='english', max_features=100000)
dtm_tfidf = tfidf_vectorizer.fit_transform(df['text'].astype(str))  # use raw text for TF-IDF

def compute_nmf_models(dtm, vectorizer, texts_tokenized, start=5, limit=15, step=1, random_state=100):
    topics_range = list(range(start, limit+1, step))
    nmf_models = []
    coherences = []
    recon_errors = []
    for k in tqdm(topics_range, desc="Training NMF models"):
        nmf = NMF(n_components=k, random_state=random_state, init='nndsvda', max_iter=400)
        W = nmf.fit_transform(dtm)
        H = nmf.components_
        nmf_models.append(nmf)
        # extract top words per topic and compute gensim coherence (c_v)
        feature_names = vectorizer.get_feature_names_out()
        topics_words = []
        for topic_idx in range(k):
            top_indices = H[topic_idx].argsort()[::-1][:20]
            topics_words.append([feature_names[i] for i in top_indices])
        cm = CoherenceModel(topics=topics_words, texts=texts_tokenized, dictionary=id2word, coherence='c_v')
        coherences.append(cm.get_coherence())
        recon_errors.append(nmf.reconstruction_err_)
    return topics_range, nmf_models, coherences, recon_errors

nmf_topics_range, nmf_models, nmf_coherences, nmf_recon_errors = compute_nmf_models(
    dtm_tfidf, tfidf_vectorizer, texts_bigrams, start=5, limit=15, step=1)
print("Best NMF coherence at k=", nmf_topics_range[int(np.argmax(nmf_coherences))], "c_v=", max(nmf_coherences))
```

```python
#  inspect best NMF model topics
best_nmf_idx = int(np.argmax(nmf_coherences))
best_nmf_k = nmf_topics_range[best_nmf_idx]
best_nmf = nmf_models[best_nmf_idx]
print(f"Selected NMF model: {best_nmf_k} topics, coherence={nmf_coherences[best_nmf_idx]:.4f}")
feature_names = tfidf_vectorizer.get_feature_names_out()
H = best_nmf.components_
for topic_idx in range(best_nmf_k):
    top_indices = H[topic_idx].argsort()[::-1][:12]
    print(f"Topic {topic_idx}: ", ", ".join(feature_names[i] for i in top_indices))
```
